```python
from pydantic iport BaseModel

class Ticker(BaseModel):
    symbol: str = "APPL"
    category: str = "EQUITY"


class Position(Basemodel):
    ticker: Ticker
    size: float
    entry_price: float

    @property
    def cost()
        return self.size * self.entry_price

class Portfolio(Basemodel):
    positions: list(Position)
    user: str

    @property
    def tot_value(self):
        return sum(p.cost for p in self.positions)
```
